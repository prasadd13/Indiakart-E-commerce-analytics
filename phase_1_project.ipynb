```python
print("---Missing values Analysis ----")

for name,df in datasets.items():
  null_counts=df.isnull().sum()
  null_pct=(df.isnull().sum()/len(df))*100

  missing_info=pd.DataFrame({
      'Missing_Count':null_counts,
      'missing_pct(%)':null_pct
  })
  only_missing=missing_info[missing_info['Missing_Count']>0]
  print(f"\n--- Table :{name.upper()}----")


  if missing_info.empty:
     print("✅ No missing values found.")
  else:
     print("Missing values found:")
     print(missing_info)
```

**Step 5: Convert Date columns to Proper Datetime Objects across all tables**

```python
orders_df ['order_date'] = pd.to_datetime(orders_df['order_date'], format="%d-%m-%Y", errors='coerce')
orders_df['delivered_date'] = pd.to_datetime(orders_df['delivered_date'], format="%d-%m-%Y", errors='coerce')
customers_df['registration_date'] = pd.to_datetime(customers_df['registration_date'], format = "%d-%m-%Y", errors='coerce')
payments_df['payment_date'] = pd.to_datetime(payments_df['payment_date'], format="%d-%m-%Y", errors='coerce')

print("=== ALL DATE COLUMNS SUCCESSFULLY CONVERTED ===")
print("orders_df -> order_date type:", orders_df['order_date'].dtype)
print("customers_df -> registration_date type:", customers_df['registration_date'].dtype)
```

**Step 7:Specific Business Logic Checks**

```python
print("=== SPECIFIC BUSINESS QUALITY CHECKS ===")

#1. Invalid Delivery Dates Check

invalid_delivery = orders_df [orders_df['delivered_date'] < orders_df['order_date']]

print(f"1. Orders with delivered_date BEFORE order_date: {len(invalid_delivery)}")

#2. Payment Gateway Failure Rate


status_col='payment_status' if 'payment_status' in payments_df.columns else 'status'

failed_payments = payments_df [payments_df [status_col] == 'Failed']

failure_rate = (len(failed_payments) / len(payments_df)) * 100

print(f"2. Payment Failure Rate: {failure_rate:.2f}% ({len(failed_payments)} failed out of {len(payments_df)})")



#3. Returns Referential Integrity Check

unmatched_returns=returns_df [~returns_df ['order_id'].isin(orders_df ['order_id'])]

print(f"3. Returns with no matching order_id in orders.csv: {len(unmatched_returns)}")

#4. Inventory Out-of-Stock Check

out_of_stock=inventory_df[inventory_df['status'] == 'Out of Stock']

print(f"4. Products in 'Out of Stock' status: {len(out_of_stock)}")

#5. Customers with Zero Orders Check

zero_order_cust=customers_df[customers_df ['total_orders'] == 0]

print(f"5. Registered customers with e total orders: {len (zero_order_cust)}")
```

## 2. Duplicate Analysis
No duplicate rows were found across any of the datasets, indicating good data integrity in terms of unique records.

## 3. Missing Values Analysis
Missing values were identified in two datasets:

*   **orders_df**: The `delivered_date` column has 17,501 missing values, accounting for approximately 35.002% of the records. This indicates a significant portion of orders where the delivery date is not recorded.
*   **payments_df**: The `refund_date` column has 50,000 missing values, which is 100% of its entries. This suggests that refunds are either not tracked, or the column is intended to be null for non-refunded payments.

Other datasets (order_items, suppliers, inventory, customers, products, returns) have no missing values.

## 4. Data Type Conversion
Date columns in `orders_df`, `customers_df`, and `payments_df` (`order_date`, `delivered_date`, `registration_date`, `payment_date`) were successfully converted to `datetime64[ns]` format. This is crucial for time-series analysis and accurate date comparisons.

## 5. Outlier Analysis (Final Amount > ₹500,000)
An outlier analysis on `orders_df` for `final_amount` revealed 312 orders with `final_amount` exceeding ₹500,000. These outliers might represent high-value orders or potential data entry errors and warrant further investigation.

## 6. Specific Business Quality Checks
Several business-specific quality checks were performed:

*   **Invalid Delivery Dates**: No orders were found where `delivered_date` was prior to `order_date`, indicating consistency in date logic.
*   **Payment Gateway Failure Rate**: The payment failure rate was calculated at **3.50%** (1,748 failed payments out of 50,000). This metric could be an area for operational improvement.
*   **Returns Referential Integrity**: All `order_id`s in the `returns_df` have a matching `order_id` in `orders_df`, confirming good referential integrity for returns data.
*   **Inventory Out-of-Stock**: 2 products were found to be 'Out of Stock' in the `inventory_df`. This highlights specific inventory management issues.
*   **Customers with Zero Orders**: 457 registered customers have a `total_orders` count of 0. This segment could be targeted for engagement campaigns or re-evaluation of registration processes.

## Summary and Recommendations
Overall, the datasets demonstrate good quality with no duplicates and successful data type conversions. However, significant missing values in `orders_df['delivered_date']` and `payments_df['refund_date']` require attention. The presence of high-value order outliers and a 3.50% payment failure rate also suggests areas for further business investigation and potential data cleaning or process improvements.
